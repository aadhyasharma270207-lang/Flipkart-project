In today's highly competitive e-commerce ecosystem, delivering exceptional customer support is vital for sustaining brand loyalty and customer retention. 1. **Data Exploration & Cleaning**:
   - Analyzed 85,907 rows across 20 features[cite: 2]. Verified zero duplicate records[cite: 2].

## 3. ***Data Wrangling***

### Data Wrangling Code

> Add blockquote



````python
# 1. Convert timestamp strings to datetime format
df['Issue_reported at'] = pd.to_datetime(df['Issue_reported at'], format='%d/%m/%Y %H:%M', errors='coerce')
df['issue_responded'] = pd.to_datetime(df['issue_responded'], format='%d/%m/%Y %H:%M', errors='coerce')

# 2. Derive response time in minutes
df['response_time_minutes'] = (df['issue_responded'] - df['Issue_reported at']).dt.total_seconds() / 60.0
df['response_time_minutes'] = df['response_time_minutes'].clip(lower=0)

# 3. Impute missing text and categorical attributes
df['Customer Remarks'] = df['Customer Remarks'].fillna('No Feedback Given')
df['Customer_City'] = df['Customer_City'].fillna('Unknown')
df['Product_category'] = df['Product_category'].fillna('Uncategorized')
df['Item_price'] = df['Item_price'].fillna(df['Item_price'].median())
df['connected_handling_time'] = df['connected_handling_time'].fillna(df['connected_handling_time'].median())

# 4. Create Binary Target Variable for CSAT Classification (1 = Satisfied [Score 4-5], 0 = Unsatisfied [Score 1-3])
df['CSAT_Satisfied'] = df['CSAT Score'].apply(lambda score: 1 if score >= 4 else 0)

print("Data Wrangling Completed Successfully!")
print(f"Updated DataFrame Shape: {df.shape}")
print("\nCSAT Binary Class Proportions (%):")
print((df['CSAT_Satisfied'].value_counts(normalize=True) * 100).round(2))
````

### What all manipulations have you done and insights you found?

#### **Data Manipulations Performed:**
1. **Datetime Conversion & Feature Derivation:** Converted `Issue_reported at` and `issue_responded` timestamps to `datetime` data types and calculated `response_time_minutes` (difference in minutes between report time and response time).
2. **Missing Value Imputation:**
   * Text/Categorical fields (`Customer Remarks`, `Customer_City`, `Product_category`) were imputed with explicit categorical tokens (`'No Feedback Given'`, `'Unknown'`, `'Uncategorized'`) to preserve pattern indicators[cite: 2].
   * Skewed numeric attributes (`Item_price`, `connected_handling_time`) were imputed using robust median values[cite: 2].
3. **Target Variable Formulation (`CSAT_Satisfied`):** Binned original discrete `CSAT Score` (1-5) into a binary target where Scores 4 and 5 indicate **Satisfied (1)** and Scores 1, 2, and 3 indicate **Unsatisfied (0)**[cite: 2].

#### **Insights Discovered:**
* **Class Distribution Imbalance:** **82.46%** of interactions resulted in satisfied outcomes, while **17.54%** resulted in dissatisfaction[cite: 2].
* **Response Time Variance:** Derived response times vary significantly, highlighting speed of resolution as a key hypothesis for customer satisfaction[cite: 2].

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

1. **Hypothetical Statement 1:** There is a statistically significant difference in mean response time (`response_time_minutes`) between satisfied (CSAT 4–5) and unsatisfied (CSAT 1–3) customers.
2. **Hypothetical Statement 2:** Customer satisfaction outcome (`CSAT_Satisfied`) is significantly dependent on the customer service channel used (`Inbound`, `Outcall`, `Email`).
3. **Hypothetical Statement 3:** Customer satisfaction outcome (`CSAT_Satisfied`) varies significantly across agent tenure experience levels (`Tenure Bucket`).

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

* **Null Hypothesis ($H_0$):** There is no significant difference in mean response time between satisfied and unsatisfied customers[cite: 2]. ($\mu_1 = \mu_2$)
* **Alternate Hypothesis ($H_1$):** Satisfied customers experience a significantly lower mean response time compared to unsatisfied customers[cite: 2]. ($\mu_1 \neq \mu_2$)

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test 1: Two-Sample Independent T-Test
from scipy import stats

satisfied_response = df[df['CSAT_Satisfied'] == 1]['response_time_minutes'].dropna()
unsatisfied_response = df[df['CSAT_Satisfied'] == 0]['response_time_minutes'].dropna()

t_stat, p_val = stats.ttest_ind(satisfied_response, unsatisfied_response, equal_var=False)

print(f"--- Hypothesis Test 1 Results ---")
print(f"T-Statistic Value : {t_stat:.4f}")
print(f"P-Value           : {p_val:.4e}")

if p_val < 0.05:
    print("Conclusion: Reject Null Hypothesis (H0). Response times differ significantly between satisfied and unsatisfied customers.")
else:
    print("Conclusion: Fail to Reject Null Hypothesis (H0).")
````

##### Which statistical test have you done to obtain P-Value?

**Two-Sample Independent T-Test (Welch's T-Test)**[cite: 2].

##### Why did you choose the specific statistical test?

Welch's T-Test was chosen because we are comparing the means of a continuous numerical variable (`response_time_minutes`) across two independent binary groups (`CSAT_Satisfied`: 0 vs 1) without assuming equal group variances[cite: 2].

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

* **Null Hypothesis ($H_0$):** Customer satisfaction outcome (`CSAT_Satisfied`) is independent of the support channel (`channel_name`)[cite: 2].
* **Alternate Hypothesis ($H_1$):** Customer satisfaction outcome (`CSAT_Satisfied`) is significantly dependent on the support channel (`channel_name`)[cite: 2].

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test 2: Chi-Square Test of Independence
contingency_channel = pd.crosstab(df['channel_name'], df['CSAT_Satisfied'])

chi2, p_val, dof, expected = stats.chi2_contingency(contingency_channel)

print(f"--- Hypothesis Test 2 Results ---")
print(f"Chi-Square Statistic : {chi2:.4f}")
print(f"Degrees of Freedom   : {dof}")
print(f"P-Value              : {p_val:.4e}")

if p_val < 0.05:
    print("Conclusion: Reject Null Hypothesis (H0). Support channel choice significantly influences CSAT outcomes.")
else:
    print("Conclusion: Fail to Reject Null Hypothesis (H0).")
````

##### Which statistical test have you done to obtain P-Value?

**Chi-Square Test of Independence ($\chi^2$)**[cite: 2].

##### Why did you choose the specific statistical test?

The Chi-Square Test of Independence is the standard non-parametric test used to determine whether a significant association exists between two categorical variables (`channel_name` vs `CSAT_Satisfied`)[cite: 2].

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

* **Null Hypothesis ($H_0$):** Customer satisfaction outcome is independent of agent experience level (`Tenure Bucket`)[cite: 2].
* **Alternate Hypothesis ($H_1$):** Customer satisfaction outcome significantly depends on agent experience level (`Tenure Bucket`)[cite: 2].

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test 3: Chi-Square Test for Tenure Buckets
contingency_tenure = pd.crosstab(df['Tenure Bucket'], df['CSAT_Satisfied'])

chi2_t, p_val_t, dof_t, expected_t = stats.chi2_contingency(contingency_tenure)

print(f"--- Hypothesis Test 3 Results ---")
print(f"Chi-Square Statistic : {chi2_t:.4f}")
print(f"Degrees of Freedom   : {dof_t}")
print(f"P-Value              : {p_val_t:.4e}")

if p_val_t < 0.05:
    print("Conclusion: Reject Null Hypothesis (H0). Agent experience level significantly impacts CSAT outcomes.")
else:
    print("Conclusion: Fail to Reject Null Hypothesis (H0).")
````

##### Which statistical test have you done to obtain P-Value?

**Chi-Square Test of Independence ($\chi^2$)**[cite: 2].

The Chi-Square test is suitable for evaluating whether proportions across multi-category ordinal features (`Tenure Bucket`) differ significantly across binary classification targets (`CSAT_Satisfied`)[cite: 2].Answer Here.
